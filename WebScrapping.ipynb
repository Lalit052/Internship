Question 4:-Write a python program to scrape first 10 product details which include product name , price , Image URL from 
https://www.bewakoof.com/bestseller?sort=popular%20

```python
url = "https://www.bewakoof.com/bestseller?sort=popular%20"
response = requests.get(url)
response
```

```python
soup=BeautifulSoup(response.content)
```

```python
name=[]
for i in soup.find_all('div',class_="productNaming bkf-ellipsis"):
    name.append(i.text)

name
```

```python
price=[]
for i in soup.find_all('span',{'class':"discountedPriceText clr-p-black   false  "}):
    price.append(str(i.text).strip())

price
```

Question 5:-Please visit https://www.cnbc.com/world/?region=world and scrap-
a) headings
b) date
c) News link

```python
site=requests.get('https://www.cnbc.com/world/?region=world')
site
```

```python
method=BeautifulSoup(site.content)
```

```python
headings=[]
for i in method.find_all('div',class_="RiverHeadline-headline RiverHeadline-hasThumbnail"):
    headings.append(i.text)

headings
```

```python
date=[]
for i in method.find_all('span',class_="RiverByline-datePublished"):
    date.append(i.text)

date
```

```python
news_link=[]
link=[]
for i in method.find_all('div', class_="RiverHeadline-headline RiverHeadline-hasThumbnail") :
    news_link.append(i)
    for x in news_link:
        for link in x.find_all('a',href=True):
            link.append(link['href'])

news_link
link
```

Question 6:- Please visit https://www.keaipublishing.com/en/journals/artificial-intelligence-in-agriculture/most-downloaded-
articles/ and scrap-
a) Paper title
b) date
c) Author

```python
page=requests.get("https://www.keaipublishing.com/en/journals/artificial-intelligence-in-agriculture/most-downloaded-articles/")
page
```

```python
soup=BeautifulSoup(page.content)
```

```python
titles=[]
for i in soup.find_all('h2',class_="h5 article-title"):
    i=i.text.replace('\n','')
    i=i.replace('\r','')
    i=i.replace('  ','')
    titles.append(i)

titles
```

```python
dates=[]
for i in soup.find_all('p',class_="article-date"):
    dates.append(i.text)

dates
```

```python
authors=[]
for i in soup.find_all('p',class_="article-authors"):
    authors.append(i.text)

authors
```
