Question 3- Create a function in python to find all words that are at least 4 characters long in a string. The use of the re.compile() method is mandatory.

```python
Text="I am Lalit and i am learning Data Science Course From DataTrained. and this is sample Text"
def find(text):
    pattern=re.compile(r'\b\w{4,}\b')
    match=pattern.findall(text)
    return match

Final_text=find(Text)
print(Final_text)
```

Question 4- Create a function in python to find all three, four, and five character words in a string. The use of the re.compile() method is mandatory.

```python
Text="I am Lalit and i am learning Data Science Course From DataTrained. and this is sample Text"
def find(text):
    pattern=re.compile(r'\b[a-zA-Z]{3,5}\b')
    match=pattern.findall(text)
    return match

Final_text=find(Text)
print(Final_text)
```

Question 5- Create a function in Python to remove the parenthesis in a list of strings. The use of the re.compile() method is mandatory

```python
string= [' "Hey(How)", "Are(You)" , "I am (Good)" ']
def remove(x):
    pattern=re.compile(r'\([^)]*\)')
    return [re.sub(pattern, '', s) for s in x]
match=remove(string)
print(match)
                   
```

Question 6- Write a python program to remove the parenthesis area from the text stored in the text file using Regular Expression.

Question 8- Create a function in python to insert spaces between words starting with numbers.
Sample Text: “RegularExpression1IsAn2ImportantTopic3InPython"

```python
sample="RegularExpression1IsAn2ImportantTopic3InPython"
def space(sample):
    pattern='(\d)([A-Z])'
    final=re.sub(pattern,r' \1\2',sample)
    return final

final_result=space(sample)
print(final_result)



```

Question 25- Write a Python program to remove continuous duplicate words from Sentence using Regular Expression.

Question 29- Write a python program to extract dates from the text stored in the text file.
Sample Text: Ron was born on 12-09-1992 and he was admitted to school 15-12-1999.
Note- Store this sample text in the file and then extract dates


```python
with open('sample.txt','w') as file:
    file.write("Ron was born on 12-09-1992 and he was admitted to school 15-12-1999.")

with open('sample.txt','r') as file:
    Text=file.read()

date=r'\d{2}-\d{2}-\d{4}'
result=re.findall(date,Text)
print(result)
```
