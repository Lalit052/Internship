Q1: In this question you have to scrape data using the filters available on the webpage You have to use the location and 
salary filter. 
You have to scrape data for “Data Scientist” designation for first 10 job results. 
You have to scrape the job-title, job-location, company name, experience required. 
The location filter to be used is “Delhi/NCR”. The salary filter to be used is “3-6” lakhs 
The task will be done as shown in the below steps: 
1. first get the web page https://www.naukri.com/
2. Enter “Data Scientist” in “Skill, Designations, and Companies” field. 
3. Then click the search button. 
4. Then apply the location filter and salary filter by checking the respective boxes 
5. Then scrape the data for the first 10 jobs results you get. 
6. Finally create a dataframe of the scraped data.

```python
import selenium
import pandas as pd 
from selenium import webdriver
import warnings
warnings.filterwarnings('ignore')
from selenium.webdriver.common.by import By
import time
```

```python
driver=webdriver.Chrome()
```

```python
#opening the page
driver.get('https://www.naukri.com/')
```

```python
#To write Data Scientist in search field
desigination=driver.find_element(By.CLASS_NAME,"suggestor-input ")
desigination.send_keys('Data Scientist')
```

```python
#to click on search
search=driver.find_element(By.CLASS_NAME,'qsbSubmit')
search.click()
```

```python
#to markdown boxes
salary=driver.find_element(By.XPATH,'/html/body/div/div/main/div[1]/div[1]/div/div/div[2]/div[4]/div[2]/div[3]/label/i')
salary.click()
```

```python
location=driver.find_element(By.XPATH,'/html/body/div/div/main/div[1]/div[1]/div/div/div[2]/div[5]/div[2]/div[2]/label/i')
location.click()
```

```python
job_title=[]
company_name=[]
Experience=[]
location=[]
```

```python

#to scrap the details of the job

title_tags=driver.find_elements(By.XPATH,'//a[@class="title "]')
for i in title_tags[0:10]:
    title=i.text
    job_title.append(title)

company_tags=driver.find_elements(By.XPATH,'//a[@class=" comp-name mw-25"]')
for i in company_tags[0:10]:
    name=i.text
    company_name.append(name)

exp_tags=driver.find_elements(By.XPATH,'//span[@class="expwdth"]')
for i in exp_tags[0:10]:
    exp=i.text
    Experience.append(exp)

location_tags=driver.find_elements(By.XPATH,'//span[@class="ni-job-tuple-icon ni-job-tuple-icon-srp-location loc"]')
for i in location_tags[0:10]:
    loc=i.text
    location.append(loc)
```

```python
print(len(job_title),len(company_name),len(Experience),len(location))
```

```python
#creating The DataFrame 
df=pd.DataFrame({'Job_Title':job_title,'Company':company_name,'Experience_Required':Experience,'Location':location})
df
```

(i.e. Car name and Price) from https://www.motor1.com/
This task will be done in following steps:
1. First get the webpage https://www.motor1.com/
2. Then You have to type in the search bar ’50 most expensive cars’
3. Then click on 50 most expensive carsin the world..
4. Then scrap thementioned data and make the dataframe

```python
driver=webdriver.Chrome()
driver.get('https://www.motor1.com/')
```

```python
input=driver.find_element(By.XPATH,'/html/body/div[9]/div[2]/div/div/div[3]/div/div/div/form/input')
input.send_keys('50 most expensive cars')
time.sleep(3)
```

```python
search=driver.find_element(By.XPATH,'/html/body/div[9]/div[2]/div/div/div[3]/div/div/div/form/button[1]')
search.click()
time.sleep(3)
```

```python
press=driver.find_element(By.XPATH,'/html/body/div[9]/div[9]/div/div[1]/div/div/div[1]/div/div[1]/h3/a')
press.click()
time.sleep(3)
```

```python
#creating empty list
car_name=[]
car_price=[]
```

```python
name=driver.find_elements(By.XPATH,'//h3[@class="subheader"]')
for i in name[0:50]:
    title=i.text
    car_name.append(title)
```

```python
#we are using //p/strong because their is no class of p tag and we have to scrap the data
price=driver.find_elements(By.XPATH,'//p/strong')
for i in price:
    money=i.text
    car_price.append(money)
```
